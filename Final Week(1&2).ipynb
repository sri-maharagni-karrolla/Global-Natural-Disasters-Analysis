```python
# DATA TRANSFORMATION
# clean column names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
print(df.columns.tolist())
```

```python
# Handle missing values
df = df.dropna(subset=["year", "country", "disaster_type"])
```

```python
# Convert numeric columns safely
numeric_cols = ["total_deaths", "total_affected", "total_damages_('000_us$)"]
```

```python
for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)
```

```python
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.preprocessing import LabelEncoder

# Select features (independent variables)
X = df[["total_deaths", "total_affected", "total_damages_('000_us$)"]]

# Target variable (dependent variable)
y = df["disaster_group"]

# Encode target labels into numeric values
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Feature selection
selector = SelectKBest(score_func=f_classif, k="all")  # "all" = keep all features
X_new = selector.fit_transform(X, y_encoded)

# Show scores
feature_scores = pd.DataFrame({
    "Feature": X.columns,
    "Score": selector.scores_
}).sort_values(by="Score", ascending=False)

print(feature_scores)
```

```python
# Drop rows with missing target
X = X.fillna(0)
y = y.dropna()
```

```python
# Select top features
selector = SelectKBest(score_func=f_classif, k="all")
fit = selector.fit(X, y)

# Show feature scores
feature_scores = pd.DataFrame({"Feature": X.columns, "Score": fit.scores_})
print(feature_scores.sort_values(by="Score", ascending=False))
```

```python
import pickle
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder

# Select features (independent variables)
X = df[["total_deaths", "total_affected", "total_damages_('000_us$)"]]

# Target variable
y = df["disaster_group"]

# Encode target
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

# Train model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Save model
with open("model.pkl", "wb") as f:
    pickle.dump(model, f)

print("✅ Model trained and saved as model.pkl")
```
